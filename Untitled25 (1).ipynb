```python
# 1. Load ONLY metadata columns
file_path = '/content/drive/MyDrive/dataset-capstone/bodmas-timestamps.parquet'
meta_cols = ['Label', 'sha256', 'family', 'category', 'timestamp']

print(f"Loading metadata columns from {file_path}...")
try:
    df = pd.read_parquet(file_path, columns=meta_cols)
    # Use format='mixed' and utc=True to handle both timezone-aware and timezone-naive strings
    df['timestamp'] = pd.to_datetime(df['timestamp'], format='mixed', utc=True)
except FileNotFoundError:
    print(f"Error: Could not find the dataset at {file_path}. Please check the path.")
    raise

print("\n--- 2. Dataset Shape and Dtypes ---")
print(f"Shape: {df.shape}")
print(f"Data Types:\n{df.dtypes}")
print("-" * 60)

# 3. Basic summary
total_rows = len(df)
benign_cnt = (df['Label'] == 0).sum()
malware_cnt = (df['Label'] == 1).sum()

malware_df = df[df['Label'] == 1].copy()
# Handle the case where 'family' might be None or NaN
malware_df = malware_df.dropna(subset=['family'])
unique_families = malware_df['family'].nunique()
unique_categories = df['category'].nunique()
category_list = df['category'].dropna().unique()

print("\n--- 3. Basic Summary ---")
print(f"Total rows: {total_rows}")
print(f"Benign (Label=0): {benign_cnt}")
print(f"Malware (Label=1): {malware_cnt}")
print(f"Unique Malware Families: {unique_families}")
print(f"Total Unique Categories: {unique_categories}")
print(f"List of Unique Categories: {list(category_list)}")
print("-" * 60)

# 4. Timestamp analysis
benign_df = df[df['Label'] == 0].copy()

benign_min_ts = benign_df['timestamp'].min()
benign_max_ts = benign_df['timestamp'].max()
malware_min_ts = malware_df['timestamp'].min()
malware_max_ts = malware_df['timestamp'].max()

print("\n--- 4. Timestamp Analysis ---")
print(f"Benign  Min Timestamp: {benign_min_ts}")
print(f"Benign  Max Timestamp: {benign_max_ts}")
print(f"Malware Min Timestamp: {malware_min_ts}")
print(f"Malware Max Timestamp: {malware_max_ts}")
print("-" * 60)

# 5. Sample counts per month
print("\n--- 5. Sample Counts Per Month ---")
# Use strftime to handle tz-aware datetimes safely for grouping and plotting
df['month'] = df['timestamp'].dt.strftime('%Y-%m')

monthly_counts = df.groupby(['month', 'Label']).size().unstack(fill_value=0)
# Rename columns for clarity
if 0 in monthly_counts.columns: monthly_counts.rename(columns={0: 'Benign'}, inplace=True)
if 1 in monthly_counts.columns: monthly_counts.rename(columns={1: 'Malware'}, inplace=True)

print(monthly_counts)

# Plotting the monthly counts
plt.figure(figsize=(12, 5))
sns.lineplot(data=monthly_counts, marker='o')
plt.title("Sample Counts Per Month")
plt.xlabel("Month")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
print("-" * 60)

# 6. Family analysis (malware only)
print("\n--- 6. Family Analysis (Top 20 Malware Families) ---")
# Apply 'month' column to malware_df as well
malware_df['month'] = malware_df['timestamp'].dt.strftime('%Y-%m')

top20_family_names = malware_df['family'].value_counts().head(20).index
top20_df = malware_df[malware_df['family'].isin(top20_family_names)]

def distinct_months_ge_5(x):
    counts = x.value_counts()
    return (counts >= 5).sum()

top20_stats = top20_df.groupby('family').agg(
    total_sample_count=('sha256', 'count'),
    earliest_timestamp=('timestamp', 'min'),
    latest_timestamp=('timestamp', 'max'),
    distinct_months_present=('month', 'nunique'),
    distinct_months_ge_5=('month', distinct_months_ge_5)
).sort_values('total_sample_count', ascending=False)

print(top20_stats.to_string())
print(f"\nTotal number of unique families: {unique_families}")
print("-" * 60)

# 7. Sequence Feasibility Check
print("\n--- 7. Sequence Feasibility Check ---")
family_stats_all = malware_df.groupby('family').agg(
    total_samples=('sha256', 'count'),
    num_months=('month', 'nunique')
)

cond1 = (family_stats_all['total_samples'] >= 5).sum()
cond2 = ((family_stats_all['total_samples'] >= 5) & (family_stats_all['num_months'] >= 3)).sum()
cond3 = ((family_stats_all['total_samples'] >= 10) & (family_stats_all['num_months'] >= 4)).sum()

print(f"Families with >= 5 total samples: {cond1}")
print(f"Families with >= 5 samples AND span >= 3 distinct months: {cond2}")
print(f"Families with >= 10 samples AND span >= 4 distinct months: {cond3}")
print("-" * 60)

# 8. Category analysis
print("\n--- 8. Category Analysis (Top 20 Categories) ---")
cat_counts = df.groupby(['category', 'Label']).size().unstack(fill_value=0)
if 0 in cat_counts.columns: cat_counts.rename(columns={0: 'Benign'}, inplace=True)
if 1 in cat_counts.columns: cat_counts.rename(columns={1: 'Malware'}, inplace=True)

cat_counts['Total'] = cat_counts.sum(axis=1)
cat_counts_sorted = cat_counts.sort_values('Total', ascending=False).head(20)
print(cat_counts_sorted)
print("-" * 60)

# 9. Plot a Gantt-style timeline for top 10 malware families
print("\n--- 9. Gantt-Style Timeline (Top 10 Malware Families) ---")
top10_families = top20_family_names[:10]
timeline_df = top20_stats.loc[top10_families].copy()
# Sort ascending for plotting top-down
timeline_df = timeline_df.sort_values('earliest_timestamp')

fig, ax = plt.subplots(figsize=(10, 6))

for idx, (family, row) in enumerate(timeline_df.iterrows()):
    start = mdates.date2num(row['earliest_timestamp'])
    end = mdates.date2num(row['latest_timestamp'])

    # Ensure a minimum visual width if start == end (e.g., samples all on one day)
    width = end - start
    if width < 1.0: width = 1.0

    ax.barh(y=family, width=width, left=start, color='coral', edgecolor='black')

ax.xaxis_date()
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
plt.xticks(rotation=45)
plt.xlabel("Timeline (Year-Month)")
plt.ylabel("Malware Family")
plt.title("Active Timeline of Top 10 Malware Families (First Seen to Last Seen)")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
print("-" * 60)

# 10. Summary Bullets
print("\n--- 10. Summary of Key Findings ---")
print(f"1. Dataset Balance: The dataset is significantly imbalanced, consisting of {benign_cnt} benign samples ({(benign_cnt/total_rows)*100:.1f}%) and {malware_cnt} malware samples ({(malware_cnt/total_rows)*100:.1f}%).")
print(f"2. Timestamp Ranges: Benign samples span from {benign_min_ts.date()} to {benign_max_ts.date()}, while malware samples span from {malware_min_ts.date()} to {malware_max_ts.date()}. Potential concept drift or differing collection periods should be evaluated.")
print(f"3. Sequence Modeling Feasibility: Out of {unique_families} malware families, {cond3} families meet the robust criteria (>= 10 samples across >= 4 months).")
if cond3 > 10:
    print("4. Conclusion: The data SUPPORTS family-level sequence modeling, but is constrained to a focused subset of high-frequency families. Most families are too sparse or short-lived.")
else:
    print("4. Conclusion: The data DOES NOT strongly support family-level sequence modeling due to extreme sparsity and short lifespans of most families.")
print("-" * 60)
```

```python
# --- PART A: DIAGNOSTICS ---
print("\n" + "="*60)
print("PART A: DIAGNOSTICS")
print("="*60)

benign_df = df[df['Label'] == 0]

# A1. Benign timestamp distribution
print("\n--- A1. Benign Timestamp Distribution ---")
print("Benign samples grouped by year:")
print(benign_df['timestamp'].dt.year.value_counts().sort_index().to_string())

cutoff_start = pd.Timestamp('2019-08-01', tz='UTC')
cutoff_end = pd.Timestamp('2020-09-30 23:59:59', tz='UTC')

old_benign = benign_df[benign_df['timestamp'] < cutoff_start]
print(f"\nBenign samples < 2019-08-01: {len(old_benign)} ({len(old_benign)/len(benign_df):.2%})")

bin_before = len(benign_df[benign_df['timestamp'] < cutoff_start])
bin_between = len(benign_df[(benign_df['timestamp'] >= cutoff_start) & (benign_df['timestamp'] <= cutoff_end)])
bin_after = len(benign_df[benign_df['timestamp'] > cutoff_end])

print("\nBenign Bins:")
print(f"  * Before 2019-08-01: {bin_before}")
print(f"  * 2019-08-01 to 2020-09-30: {bin_between}")
print(f"  * After 2020-09-30: {bin_after}")

# A2. Mismatch subset 1
print("\n--- A2. Mismatch Subset 1 (Label=1 AND category='benign') ---")
mismatch1 = df[(df['Label'] == 1) & (df['category'] == 'benign')]
print(f"Count: {len(mismatch1)}")

if not mismatch1.empty:
    sample1 = mismatch1.head(10).copy()
    sample1['sha256'] = sample1['sha256'].str[:16] + "..."
    print("\nSample rows:")
    print(sample1[['sha256', 'family', 'timestamp', 'category']].to_string(index=False))

    none_family1 = mismatch1['family'].isna().sum()
    real_family1 = len(mismatch1) - none_family1
    print(f"\nFamily IS None: {none_family1}")
    print(f"Real family name: {real_family1}")
    print(f"Min timestamp: {mismatch1['timestamp'].min()}")
    print(f"Max timestamp: {mismatch1['timestamp'].max()}")

# A3. Mismatch subset 2
print("\n--- A3. Mismatch Subset 2 (Label=0 AND category='trojan') ---")
mismatch2 = df[(df['Label'] == 0) & (df['category'] == 'trojan')]
print(f"Count: {len(mismatch2)}")

if not mismatch2.empty:
    sample2 = mismatch2.head(10).copy()
    sample2['sha256'] = sample2['sha256'].str[:16] + "..."
    print("\nSample rows:")
    print(sample2[['sha256', 'family', 'timestamp', 'category']].to_string(index=False))

    none_family2 = mismatch2['family'].isna().sum()
    real_family2 = len(mismatch2) - none_family2
    print(f"\nFamily IS None: {none_family2}")
    print(f"Real family name: {real_family2}")
    print(f"Min timestamp: {mismatch2['timestamp'].min()}")
    print(f"Max timestamp: {mismatch2['timestamp'].max()}")

# A4. Cross-tabulate Label vs Category
print("\n--- A4. Cross-tabulate Label vs Category (Top 10 Categories) ---")
top10_categories = df['category'].value_counts().head(10).index
crosstab = pd.crosstab(df['Label'], df['category'])[top10_categories]
print(crosstab.to_string())


# --- PART B: FILTERING ---
print("\n" + "="*60)
print("PART B: FILTERING")
print("="*60)

# B1. Timestamp filter
print("\n--- B1. Timestamp Filter ---")
orig_benign = (df['Label'] == 0).sum()
orig_malware = (df['Label'] == 1).sum()

df = df[(df['timestamp'] >= cutoff_start) & (df['timestamp'] <= cutoff_end)]

new_benign = (df['Label'] == 0).sum()
new_malware = (df['Label'] == 1).sum()

print(f"Dropped Benign: {orig_benign - new_benign}")
print(f"Dropped Malware: {orig_malware - new_malware}")
print(f"New Class Counts -> Benign: {new_benign}, Malware: {new_malware}")

# B2. Label/category mismatch filter
print("\n--- B2. Label/Category Mismatch Filter ---")
rule1_mask = (df['Label'] == 1) & (df['category'] == 'benign') & (df['family'].isna())
rule1_dropped = rule1_mask.sum()
df = df[~rule1_mask]

rule2_mask = (df['Label'] == 1) & (df['category'] == 'benign') & (df['family'].notna())
rule2_kept = rule2_mask.sum()

rule3_mask = (df['Label'] == 0) & (df['category'] == 'trojan')
rule3_dropped = rule3_mask.sum()
df = df[~rule3_mask]

print(f"Rule 1 (Label=1, category='benign', family=None)    -> DROPPED: {rule1_dropped}")
print(f"Rule 2 (Label=1, category='benign', family!=None)   -> KEPT: {rule2_kept}")
print(f"Rule 3 (Label=0, category='trojan')                 -> DROPPED: {rule3_dropped}")

# B3. Drop category column
df = df.drop(columns=['category'])
print("\n--- B3. Dropped 'category' column ---")

# B4. Sanity check post-filter
print("\n--- B4. Sanity Check Post-Filter ---")
final_benign = (df['Label'] == 0).sum()
final_malware = (df['Label'] == 1).sum()
print(f"Final Row Count: {len(df)}")
print(f"Final Benign: {final_benign}")
print(f"Final Malware: {final_malware}")

malware_df = df[df['Label'] == 1].copy()
unique_fams = malware_df['family'].nunique()
print(f"Final Unique Malware Families: {unique_fams}")

print(f"\nMin Timestamp (Benign):  {df[df['Label'] == 0]['timestamp'].min()}")
print(f"Max Timestamp (Benign):  {df[df['Label'] == 0]['timestamp'].max()}")
print(f"Min Timestamp (Malware): {df[df['Label'] == 1]['timestamp'].min()}")
print(f"Max Timestamp (Malware): {df[df['Label'] == 1]['timestamp'].max()}")

malware_df['month'] = malware_df['timestamp'].dt.strftime('%Y-%m')
family_stats = malware_df.groupby('family').agg(
    total_samples=('sha256', 'count'),
    num_months=('month', 'nunique')
)
robust_families = ((family_stats['total_samples'] >= 10) & (family_stats['num_months'] >= 4)).sum()
print(f"\nFamilies with >= 10 samples AND span >= 4 distinct months: {robust_families}")


# --- PART C: PREPARE FOR JOIN ---
print("\n" + "="*60)
print("PART C: PREPARATION FOR NEXT STEP")
print("="*60)

metadata_out = '/content/drive/MyDrive/dataset-capstone/bodmas_clean_metadata.parquet'
sha256_out = '/content/drive/MyDrive/dataset-capstone/bodmas_clean_sha256.txt'

# C1. Save cleaned parquet
df[['sha256', 'Label', 'family', 'timestamp']].to_parquet(metadata_out, index=False)
print(f"C1. Saved cleaned metadata to: {metadata_out}")

# C2. Save sha256 list
with open(sha256_out, 'w') as f:
    for sha in df['sha256']:
        f.write(f"{sha}\n")
print(f"C2. Saved clean SHA256 list to: {sha256_out}")

# C3. Confirmation Note
print("\nC3. NOTE: The files are now ready! In the next step, we can load 'bodmas.parquet' ")
print("and safely merge/filter it using 'bodmas_clean_sha256.txt' or the clean metadata parquet.")
print("This will ensure we only load the 1.2GB feature set for valid rows, avoiding memory crashes.")
```

```python
!pip install duckdb

import gc
import duckdb
import pandas as pd

# 1. Load the clean sha256 list from the text file into a Python set
sha256_txt_path = '/content/drive/MyDrive/dataset-capstone/bodmas_clean_sha256.txt'
with open(sha256_txt_path, 'r') as f:
    clean_sha256_set = set(line.strip() for line in f if line.strip())

print(f"1. Loaded {len(clean_sha256_set):,} SHA256 hashes into clean_sha256_set.")

# 2. Load the clean metadata parquet into a pandas DataFrame
metadata_parquet_path = '/content/drive/MyDrive/dataset-capstone/bodmas_clean_metadata.parquet'
df_meta = pd.read_parquet(metadata_parquet_path)
print(f"2. Loaded clean metadata. Shape: {df_meta.shape}")

# 3. Query the 1.2 GB feature file using DuckDB to filter on RAM efficiently
feature_parquet_path = '/content/drive/MyDrive/dataset-capstone/bodmas.parquet'

con = duckdb.connect()
# To allow DuckDB to query our list of hashes efficiently, we wrap it in a DataFrame
clean_hashes_df = pd.DataFrame({'sha256': list(clean_sha256_set)})

print("3. Querying 1.2 GB feature file via DuckDB...")
query = f"""
    SELECT *
    FROM '{feature_parquet_path}'
    WHERE sha256 IN (SELECT sha256 FROM clean_hashes_df)
"""
features_df = con.execute(query).df()
con.close()

# 4. Print the shape of features_df
print(f"\n4. Shape of features_df: {features_df.shape}")

# 5. Verify columns
cols = features_df.columns.tolist()
print("\n5. Column Verification:")
print(f"   First 5 columns: {cols[:5]}")
print(f"   Last 5 columns:  {cols[-5:]}")
print(f"   'sha256' present in features_df: {'sha256' in cols}")

# 6. Merge features_df with clean metadata on 'sha256'
# Drop uncleaned metadata columns from features_df to preserve cleaned metadata from df_meta
meta_cols_to_drop = [c for c in ['Label', 'family', 'category'] if c in features_df.columns]
features_df_subset = features_df.drop(columns=meta_cols_to_drop)

df_full = pd.merge(df_meta, features_df_subset, on='sha256', how='inner')

print(f"\n6. Shape of df_full: {df_full.shape}")

# Confirm time alignment
benign_ts = df_full[df_full['Label'] == 0]['timestamp']
malware_ts = df_full[df_full['Label'] == 1]['timestamp']

print(f"   Benign Min Timestamp:  {benign_ts.min()}")
print(f"   Benign Max Timestamp:  {benign_ts.max()}")
print(f"   Malware Min Timestamp: {malware_ts.min()}")
print(f"   Malware Max Timestamp: {malware_ts.max()}")

# Print memory usage in MB
mem_mb = df_full.memory_usage(deep=True).sum() / (1024 ** 2)
print(f"\n   Memory usage of df_full: {mem_mb:.2f} MB")

# 7. Save df_full as parquet
output_parquet_path = '/content/drive/MyDrive/dataset-capstone/bodmas_clean_features.parquet'
df_full.to_parquet(output_parquet_path, index=False)
print(f"7. Saved df_full successfully to: {output_parquet_path}")

# 8. Clean up RAM
del features_df
del features_df_subset
del clean_hashes_df
gc.collect()
print("\n8. RAM cleanup completed (`features_df` deleted and `gc.collect()` executed).")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import gc
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import roc_auc_score

import warnings
warnings.filterwarnings('ignore')

print("\n" + "="*60)
print("PART A: LOAD AND PREPARE")
print("="*60)

# A1. Load parquet
file_path = '/content/drive/MyDrive/dataset-capstone/bodmas_clean_features.parquet'
print("A1. Loading dataset...")
df = pd.read_parquet(file_path)
mem_mb = df.memory_usage(deep=True).sum() / (1024 ** 2)
print(f"    Shape: {df.shape}")
print(f"    Memory usage: {mem_mb:.2f} MB")

# A2. Convert timestamp to datetime
# Parquet likely retained UTC tz, but we'll ensure it just in case
if df['timestamp'].dt.tz is None:
    df['timestamp'] = df['timestamp'].dt.tz_localize('UTC')
else:
    df['timestamp'] = df['timestamp'].dt.tz_convert('UTC')
print("A2. Timestamp converted/verified as UTC.")

# A3. Create 'month' column as Period
df['month'] = df['timestamp'].dt.to_period('M')
print("A3. 'month' column created.")

# A4. Samples per month, split by Label
print("\nA4. Samples per month:")
monthly_counts = df.groupby(['month', 'Label']).size().unstack(fill_value=0)
if 0 in monthly_counts.columns: monthly_counts.rename(columns={0: 'Benign'}, inplace=True)
if 1 in monthly_counts.columns: monthly_counts.rename(columns={1: 'Malware'}, inplace=True)
print(monthly_counts.to_string())

# A5. Define Train, Val, Test masks
t_start_train, t_end_train = pd.Timestamp('2019-08-01', tz='UTC'), pd.Timestamp('2020-03-31 23:59:59', tz='UTC')
t_start_val, t_end_val     = pd.Timestamp('2020-04-01', tz='UTC'), pd.Timestamp('2020-05-31 23:59:59', tz='UTC')
t_start_test, t_end_test   = pd.Timestamp('2020-06-01', tz='UTC'), pd.Timestamp('2020-09-30 23:59:59', tz='UTC')

train_mask = (df['timestamp'] >= t_start_train) & (df['timestamp'] <= t_end_train)
val_mask   = (df['timestamp'] >= t_start_val) & (df['timestamp'] <= t_end_val)
test_mask  = (df['timestamp'] >= t_start_test) & (df['timestamp'] <= t_end_test)

print("\nA5. Split Summary:")
print(f"    Train (Aug 2019 - Mar 2020): {train_mask.sum()} rows")
print(f"    Val   (Apr 2020 - May 2020): {val_mask.sum()} rows")
print(f"    Test  (Jun 2020 - Sep 2020): {test_mask.sum()} rows")


print("\n" + "="*60)
print("PART B: FEATURE DRIFT PLOTS")
print("="*60)

# B1. Pick 6 features
# Since columns are F1...F2381 and don't contain textual names like 'entropy', we use the fallback list.
drift_features = ['F1', 'F50', 'F200', 'F800', 'F1500', 'F2381']
print(f"B1. Selected 6 features for plotting: {drift_features}")

# B2. Plot monthly mean vs month for the 6 features
months_str = df['month'].dt.strftime('%Y-%m').unique()
months_str.sort()
# Find the x-index for March 2020 to place the vertical line
march_idx = np.where(months_str == '2020-03')[0]
march_idx = march_idx[0] if len(march_idx) > 0 else None

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

df['month_str'] = df['month'].dt.strftime('%Y-%m')

for idx, f in enumerate(drift_features):
    ax = axes[idx]
    sns.lineplot(data=df, x='month_str', y=f, hue='Label', estimator='mean', errorbar=None,
                 marker='o', ax=ax, palette={0: 'blue', 1: 'orange'})
    ax.set_title(f"Mean {f} Over Time")
    ax.set_xlabel("Month")
    ax.set_ylabel(f"Mean {f}")
    ax.tick_params(axis='x', rotation=45)

    if march_idx is not None:
        ax.axvline(x=march_idx, color='red', linestyle='--', alpha=0.7, label='COVID Shift (2020-03)')

    if idx == 0:
        handles, labels = ax.get_legend_handles_labels()
        ax.legend(handles, ['Benign', 'Malware', 'COVID Shift'])
    else:
        ax.get_legend().remove()

plt.tight_layout()
plt.show()

# B3. Compute drift magnitude for all F1-F2381
print("\nB3. Computing drift magnitude for all 2,381 features...")
feature_cols = [c for c in df.columns if c.startswith('F') and c[1:].isdigit()]

train_df = df[train_mask]
test_df = df[test_mask]

# Compute means and standard deviations
mean_train = train_df[feature_cols].mean(numeric_only=True)
mean_test = test_df[feature_cols].mean(numeric_only=True)
std_all = df[feature_cols].std(numeric_only=True)

# Compute magnitude
drift_magnitude = (mean_test - mean_train).abs() / (std_all + 1e-8)
drift_magnitude = drift_magnitude.sort_values(ascending=False)

top20_drift = drift_magnitude.head(20)
print("\nTop 20 Drifting Features:")
print(top20_drift.to_frame(name='Drift Magnitude'))

csv_out = '/content/drive/MyDrive/dataset-capstone/feature_drift_ranking.csv'
drift_magnitude.to_csv(csv_out, header=['drift_magnitude'])
print(f"\nSaved feature drift ranking to: {csv_out}")

# B4. Plot Horizontal bar chart of top 20 drifting features
plt.figure(figsize=(10, 6))
sns.barplot(x=top20_drift.values, y=top20_drift.index, palette='viridis')
plt.title("Top 20 Features by Concept Drift Magnitude")
plt.xlabel("Drift Magnitude: |Mean Test - Mean Train| / Std Dev")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()


print("\n" + "="*60)
print("PART C: BASELINE CLASSIFIER + AUROC DECAY")
print("="*60)

# C1. Prepare X and y
print("C1. Preparing features and handling NaNs (filling with 0 for X)...")
X_all = df[feature_cols].fillna(0).astype(np.float32).values
y_all = df['Label'].astype(int).values

X_train, y_train = X_all[train_mask], y_all[train_mask]
X_val, y_val     = X_all[val_mask], y_all[val_mask]
X_test, y_test   = X_all[test_mask], y_all[test_mask]

# C2. Train baseline model
print("C2. Training HistGradientBoostingClassifier on TRAIN split only...")
clf = HistGradientBoostingClassifier(
    max_iter=200,          # akin to n_estimators=200
    max_depth=6,
    learning_rate=0.1,
    random_state=42
)
clf.fit(X_train, y_train)

# C3. Evaluate AUROC
def get_auc(X, y):
    if len(np.unique(y)) > 1:
        probs = clf.predict_proba(X)[:, 1]
        return roc_auc_score(y, probs)
    return np.nan

print("\nC3. Evaluating AUROC...")
val_auc = get_auc(X_val, y_val)
test_auc = get_auc(X_test, y_test)

print(f"    Full Val AUROC (Apr-May): {val_auc:.4f}")
print(f"    Full Test AUROC (Jun-Sep): {test_auc:.4f}")

# Monthly AUROC calculation
months_ordered = df['month'].sort_values().unique()
monthly_aucs = {}

print("\n    Individual Month AUROCs:")
for m in months_ordered:
    mask_m = df['month'] == m
    X_m = X_all[mask_m]
    y_m = y_all[mask_m]
    auc_m = get_auc(X_m, y_m)
    monthly_aucs[m] = auc_m
    m_str = m.strftime('%Y-%m')
    split = "Train" if m_str < '2020-04' else "Val" if m_str < '2020-06' else "Test"
    print(f"      {m_str} ({split}): {auc_m:.4f}")

# C4. Plot AUROC vs month
auc_df = pd.DataFrame({'month_str': [m.strftime('%Y-%m') for m in months_ordered],
                       'AUROC': [monthly_aucs[m] for m in months_ordered]})

plt.figure(figsize=(12, 6))
sns.lineplot(data=auc_df, x='month_str', y='AUROC', marker='o', color='purple', linewidth=2)

# Shade regions
train_idx_start = auc_df.index[auc_df['month_str'] == '2019-08'][0]
train_idx_end   = auc_df.index[auc_df['month_str'] == '2020-03'][0]
val_idx_start   = auc_df.index[auc_df['month_str'] == '2020-04'][0]
val_idx_end     = auc_df.index[auc_df['month_str'] == '2020-05'][0]
test_idx_start  = auc_df.index[auc_df['month_str'] == '2020-06'][0]
test_idx_end    = auc_df.index[auc_df['month_str'] == '2020-09'][0]

plt.axvspan(train_idx_start - 0.5, train_idx_end + 0.5, color='lightblue', alpha=0.3, label='Train')
plt.axvspan(val_idx_start - 0.5, val_idx_end + 0.5, color='lightyellow', alpha=0.4, label='Val')
plt.axvspan(test_idx_start - 0.5, test_idx_end + 0.5, color='lightcoral', alpha=0.3, label='Test')

plt.axhline(0.99, color='black', linestyle='--', label='Target (0.99)')
plt.xticks(rotation=45)
plt.title("Static XGBoost AUROC Decay Over Time (Concept Drift Evidence)")
plt.xlabel("Month")
plt.ylabel("AUROC")
plt.legend(loc='lower left')
plt.grid(True, linestyle=':', alpha=0.7)
plt.tight_layout()
plt.show()

# C5. Print Summary
print("\nC5. Performance Summary:")
train_aucs = [monthly_aucs[m] for m in months_ordered if m.strftime('%Y-%m') <= '2020-03']
test_aucs  = [monthly_aucs[m] for m in months_ordered if m.strftime('%Y-%m') >= '2020-06']

mean_train_auc = np.mean(train_aucs)
mean_test_auc = np.mean(test_aucs)
auroc_drop = mean_train_auc - mean_test_auc

print(f"    Mean Train AUROC: {mean_train_auc:.4f}")
print(f"    Mean Test AUROC:  {mean_test_auc:.4f}")
print(f"    AUROC Drop:       {auroc_drop:.4f}")

print("\n    VERDICT:")
verdict = ""
if auroc_drop > 0.05:
    verdict = "CONCEPT DRIFT IS SIGNIFICANT. NJODE PROPOSAL JUSTIFIED."
else:
    verdict = "DRIFT IS WEAK. RECONSIDER MODEL COMPLEXITY."
print(f"    *** {verdict} ***")


print("\n" + "="*60)
print("PART D: QUICK CLASS SEPARABILITY HEATMAP")
print("="*60)

# D1. Bin the 2,381 features into 20 groups of ~119 features each
num_groups = 20
feature_groups = np.array_split(feature_cols, num_groups)

# D2. Compute the mean |mean_malware - mean_benign| per month for each group
heatmap_data = []

months_str_list = [m.strftime('%Y-%m') for m in months_ordered]

for g_idx, group_cols in enumerate(feature_groups):
    group_monthly_separation = []

    for m in months_str_list:
        mask_m = df['month_str'] == m
        df_m = df[mask_m]

        malware_m = df_m[df_m['Label'] == 1][group_cols].mean(numeric_only=True)
        benign_m = df_m[df_m['Label'] == 0][group_cols].mean(numeric_only=True)

        # Absolute difference for each feature in the group, then mean over the group
        sep = (malware_m - benign_m).abs().mean()
        group_monthly_separation.append(sep)

    heatmap_data.append(group_monthly_separation)

heatmap_df = pd.DataFrame(heatmap_data, columns=months_str_list, index=[f"Group {i+1}" for i in range(num_groups)])

# Normalize row-wise (MinMax scaling per group to see *when* it drifts most)
heatmap_df_norm = heatmap_df.apply(lambda x: (x - x.min()) / (x.max() - x.min() + 1e-8), axis=1)

# D3. Plot Heatmap
plt.figure(figsize=(14, 8))
sns.heatmap(heatmap_df_norm, cmap="Reds", cbar_kws={'label': 'Normalized Separation Magnitude'})
plt.title("Feature Separability (Malware vs Benign) Over Time per Feature Group")
plt.xlabel("Month")
plt.ylabel("Feature Groups (119 Features per Group)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


print("\n" + "="*60)
print("PART E: SAVE FINDINGS")
print("="*60)

summary_out = '/content/drive/MyDrive/dataset-capstone/drift_eda_summary.txt'
with open(summary_out, 'w') as f:
    f.write("=== Feature Drift EDA Summary ===\n\n")
    f.write(f"Train rows (Aug 19 - Mar 20): {train_mask.sum()}\n")
    f.write(f"Val rows (Apr 20 - May 20):   {val_mask.sum()}\n")
    f.write(f"Test rows (Jun 20 - Sep 20):  {test_mask.sum()}\n\n")

    f.write("Top 20 Drifting Features:\n")
    f.write(top20_drift.to_string() + "\n\n")

    f.write("AUROC Numbers:\n")
    f.write(f"Mean Train AUROC: {mean_train_auc:.4f}\n")
    f.write(f"Full Val AUROC:   {val_auc:.4f}\n")
    f.write(f"Full Test AUROC:  {test_auc:.4f}\n")
    f.write(f"Mean Test AUROC:  {mean_test_auc:.4f}\n")
    f.write(f"AUROC Drop (Train - Test): {auroc_drop:.4f}\n\n")

    f.write("Verdict:\n")
    f.write(f"{verdict}\n")

print(f"E1. Saved summary to: {summary_out}")

# Clean up memory
del X_all, y_all, X_train, y_train, X_val, y_val, X_test, y_test
gc.collect()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import gc
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import roc_auc_score
import warnings

warnings.filterwarnings('ignore')

print("\n" + "="*60)
print("PART A: IDENTIFY NOVEL VS PERSISTENT FAMILIES")
print("="*60)

# A1. Load dataset
file_path = '/content/drive/MyDrive/dataset-capstone/bodmas_clean_features.parquet'
print("A1. Loading dataset...")
df = pd.read_parquet(file_path)

if df['timestamp'].dt.tz is None:
    df['timestamp'] = df['timestamp'].dt.tz_localize('UTC')
else:
    df['timestamp'] = df['timestamp'].dt.tz_convert('UTC')

# A2 & A3. Define windows
t_start_train, t_end_train = pd.Timestamp('2019-08-01', tz='UTC'), pd.Timestamp('2020-03-31 23:59:59', tz='UTC')
t_start_test, t_end_test   = pd.Timestamp('2020-06-01', tz='UTC'), pd.Timestamp('2020-09-30 23:59:59', tz='UTC')

train_window_mask = (df['timestamp'] >= t_start_train) & (df['timestamp'] <= t_end_train)
test_window_mask  = (df['timestamp'] >= t_start_test) & (df['timestamp'] <= t_end_test)

# A4 & A5. Extract Malware subsets and unique family sets
malware_df = df[df['Label'] == 1]
train_malware = malware_df[(malware_df['timestamp'] >= t_start_train) & (malware_df['timestamp'] <= t_end_train)]
test_malware  = malware_df[(malware_df['timestamp'] >= t_start_test) & (malware_df['timestamp'] <= t_end_test)]

train_families = set(train_malware['family'].dropna().unique())
test_families  = set(test_malware['family'].dropna().unique())

# A6 & A7. Novel vs Persistent
novel_families      = test_families - train_families
persistent_families = test_families & train_families

novel_test_malware      = test_malware[test_malware['family'].isin(novel_families)]
persistent_test_malware = test_malware[test_malware['family'].isin(persistent_families)]

# A8. Summary prints
print(f"\nNumber of Novel Families:               {len(novel_families)}")
print(f"Total Test Samples (Novel Families):     {len(novel_test_malware)}")
print(f"Number of Persistent Families:          {len(persistent_families)}")
print(f"Total Test Samples (Persistent Families): {len(persistent_test_malware)}")

print("\nTop 15 Novel Families by Sample Count:")
top15_novel = novel_test_malware['family'].value_counts().head(15)
print(top15_novel.to_string())

print("\n" + "="*60)
print("PART B: CONSTRUCT THE SPLITS")
print("="*60)

# B1. Train Split
train_df = df[train_window_mask]

# B2 & B3. Test Splits (both include test window benign samples)
test_benign = df[test_window_mask & (df['Label'] == 0)]

test_persistent_df = pd.concat([test_benign, persistent_test_malware])
test_novel_df      = pd.concat([test_benign, novel_test_malware])

# B4. Class Balance Summary
def print_split_info(name, df_split):
    b_cnt = (df_split['Label'] == 0).sum()
    m_cnt = (df_split['Label'] == 1).sum()
    print(f"{name:18s} -> Total: {len(df_split):6d} | Benign: {b_cnt:6d} | Malware: {m_cnt:6d}")

print_split_info("Train Split", train_df)
print_split_info("Test_Persistent", test_persistent_df)
print_split_info("Test_Novel", test_novel_df)

print("\n" + "="*60)
print("PART C: TRAIN AND EVALUATE BASELINE")
print("="*60)

# C1. Prepare Features & Labels
feature_cols = [c for c in df.columns if c.startswith('F') and c[1:].isdigit()]

X_train = train_df[feature_cols].fillna(0).astype(np.float32).values
y_train = train_df['Label'].astype(int).values

X_test_p = test_persistent_df[feature_cols].fillna(0).astype(np.float32).values
y_test_p = test_persistent_df['Label'].astype(int).values

X_test_n = test_novel_df[feature_cols].fillna(0).astype(np.float32).values
y_test_n = test_novel_df['Label'].astype(int).values

# C2. Train Baseline Classifier
print("Training HistGradientBoostingClassifier on Train Split...")
clf = HistGradientBoostingClassifier(
    max_iter=200,
    max_depth=6,
    learning_rate=0.1,
    random_state=42
)
clf.fit(X_train, y_train)

# C3 & C4. AUROC Evaluation & Gap
probs_p = clf.predict_proba(X_test_p)[:, 1]
probs_n = clf.predict_proba(X_test_n)[:, 1]

auroc_persistent = roc_auc_score(y_test_p, probs_p)
auroc_novel      = roc_auc_score(y_test_n, probs_n)
auroc_gap        = auroc_persistent - auroc_novel

print(f"\nAUROC on Test_Persistent: {auroc_persistent:.4f}")
print(f"AUROC on Test_Novel:      {auroc_novel:.4f}")
print(f"AUROC Gap (Persistent - Novel): {auroc_gap:.4f}")

print("\n" + "="*60)
print("PART D: PER-FAMILY AUROC")
print("="*60)

top10_novel_fams = novel_test_malware['family'].value_counts().head(10).index.tolist()
top5_pers_fams   = persistent_test_malware['family'].value_counts().head(5).index.tolist()

# D1. Top 10 Novel Families
novel_family_results = []
print("Top 10 Novel Families AUROC:")
print(f"{'Family':<20} | {'Samples':<8} | {'AUROC':<8}")
print("-" * 42)
for fam in top10_novel_fams:
    fam_malware = test_malware[test_malware['family'] == fam]
    mini_test = pd.concat([test_benign, fam_malware])

    X_mini = mini_test[feature_cols].fillna(0).astype(np.float32).values
    y_mini = mini_test['Label'].astype(int).values

    probs_mini = clf.predict_proba(X_mini)[:, 1]
    auc_mini = roc_auc_score(y_mini, probs_mini)

    novel_family_results.append({'family': fam, 'samples': len(fam_malware), 'auroc': auc_mini})
    print(f"{fam:<20} | {len(fam_malware):<8d} | {auc_mini:.4f}")

# D2. Top 5 Persistent Families
pers_family_results = []
print("\nTop 5 Persistent Families AUROC:")
print(f"{'Family':<20} | {'Samples':<8} | {'AUROC':<8}")
print("-" * 42)
for fam in top5_pers_fams:
    fam_malware = test_malware[test_malware['family'] == fam]
    mini_test = pd.concat([test_benign, fam_malware])

    X_mini = mini_test[feature_cols].fillna(0).astype(np.float32).values
    y_mini = mini_test['Label'].astype(int).values

    probs_mini = clf.predict_proba(X_mini)[:, 1]
    auc_mini = roc_auc_score(y_mini, probs_mini)

    pers_family_results.append({'family': fam, 'samples': len(fam_malware), 'auroc': auc_mini})
    print(f"{fam:<20} | {len(fam_malware):<8d} | {auc_mini:.4f}")

print("\n" + "="*60)
print("PART E: VISUALIZATION & SAVE FINDINGS")
print("="*60)

# E1. Bar Chart 1: Test_Persistent vs Test_Novel AUROC
plt.figure(figsize=(7, 5))
bars = plt.bar(['Test_Persistent', 'Test_Novel'], [auroc_persistent, auroc_novel], color=['#2ca02c', '#d62728'], width=0.45)
plt.ylim(0.0, 1.0)
plt.ylabel("AUROC")
plt.title("Classifier Generalization: Persistent vs. Novel Malware Families")
plt.grid(axis='y', linestyle='--', alpha=0.7)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.02, f"{yval:.4f}", ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()

# E2. Bar Chart 2: Per-family AUROC for Top 10 Novel Families
df_novel_res = pd.DataFrame(novel_family_results).sort_values('auroc', ascending=True)

plt.figure(figsize=(10, 6))
bars2 = plt.barh(df_novel_res['family'], df_novel_res['auroc'], color='coral', edgecolor='black')
plt.xlim(0.0, 1.0)
plt.xlabel("AUROC")
plt.ylabel("Novel Malware Family")
plt.title("Per-Family AUROC for Top 10 Novel Malware Families")
plt.grid(axis='x', linestyle='--', alpha=0.7)

for bar in bars2:
    xval = bar.get_width()
    plt.text(xval + 0.01, bar.get_y() + bar.get_height()/2.0, f"{xval:.4f}", ha='left', va='center', fontsize=9)

plt.tight_layout()
plt.show()

# E3. Save Summary Text File
summary_path = '/content/drive/MyDrive/dataset-capstone/novel_family_analysis.txt'
with open(summary_path, 'w') as f:
    f.write("=== NOVEL VS PERSISTENT MALWARE FAMILY ANALYSIS ===\n\n")
    f.write(f"Novel Families Count:      {len(novel_families)}\n")
    f.write(f"Novel Family Test Samples:  {len(novel_test_malware)}\n")
    f.write(f"Persistent Families Count: {len(persistent_families)}\n")
    f.write(f"Persistent Test Samples:   {len(persistent_test_malware)}\n\n")

    f.write("SPLIT SIZES:\n")
    f.write(f"Train Split:     {len(train_df)} rows\n")
    f.write(f"Test_Persistent: {len(test_persistent_df)} rows\n")
    f.write(f"Test_Novel:      {len(test_novel_df)} rows\n\n")

    f.write("CLASSIFIER PERFORMANCE:\n")
    f.write(f"AUROC (Test_Persistent): {auroc_persistent:.4f}\n")
    f.write(f"AUROC (Test_Novel):      {auroc_novel:.4f}\n")
    f.write(f"AUROC Performance Gap:   {auroc_gap:.4f}\n\n")

    f.write("TOP 10 NOVEL FAMILIES AUROC:\n")
    for res in novel_family_results:
        f.write(f"  - {res['family']:<20}: Samples={res['samples']:<5} | AUROC={res['auroc']:.4f}\n")

    f.write("\nTOP 5 PERSISTENT FAMILIES AUROC:\n")
    for res in pers_family_results:
        f.write(f"  - {res['family']:<20}: Samples={res['samples']:<5} | AUROC={res['auroc']:.4f}\n")

print(f"\nE3. Summary successfully saved to: {summary_path}")

# RAM Cleanup
del X_train, y_train, X_test_p, y_test_p, X_test_n, y_test_n
gc.collect()
```
